```python
import tensorflow as tf
```

### __ZERO DIMENSIONAL TENSOR__

```python
tensor_zero_d = tf.constant(4) #zero dim tensor
tensor_zero_d
```

### __ONE DIMENSIONAL TENSOR__

```python
tensor_one_d = tf.constant([2,0,-3], dtype=tf.float32) #one dim tensor
tensor_one_d
```

### __TWO DIMENSIONAL TENSOR__

```python
tensor_two_d = tf.constant([
    [2,0,-3],
    [3,5,1],                     #two dim tensor
    [-9,6,7]
    ]) 
tensor_two_d
```

### __THREE DIMENSIONAL TENSOR__

```python
tensor_three_d = tf.constant([

    [
        [2,5,1],[6,7,9]
    ],
    [
        [-9,2,-7],[8,2,6]            #three dim tensor
    ],
    [
        [6,5,2],[-1,0,3]
    ]
])
tensor_three_d
```

### __FOUR DIMMENSIONAL TENSOR__

```python
tensor_four_d = tf.constant([                #four dim tensor
     [[
        [2,5,1],[6,7,9]
    ],
    [
        [-9,2,-7],[8,2,6]            #three dim tensor
    ],
    [
        [6,5,2],[-1,0,3]
    ]],
    [[
        [2,8,1],[-6,7,9]
    ],
    [
        [-2,1,-7],[5,2,6]            #three dim tensor
    ],
    [
        [1,5,3],[-1,8,3]
    ]]
])
tensor_four_d
```

### __SHAPE__

```python
print(tensor_zero_d.shape)
print(tensor_one_d.shape)
print(tensor_two_d.shape)
print(tensor_three_d.shape)
print(tensor_four_d.shape)
```

### __NDIM OF TENSOR__

```python
print(tensor_zero_d.ndim)
print(tensor_one_d.ndim)
print(tensor_two_d.ndim)
print(tensor_three_d.ndim)
print(tensor_four_d.ndim)
```

### __ABS__

### __DIVIDE__

```python
x1 = tf.constant([2,3,4,5,6],dtype=tf.float32)
x2 = tf.constant([2,0,4,0,7],dtype=tf.float32) 
tf.math.divide(x1,x2)
```

### __DIVIDE NO NAN__

```python
x1 = tf.constant([2,3,4,5,6],dtype=tf.float32)
x2 = tf.constant([2,0,4,0,7],dtype=tf.float32) 
tf.math.divide_no_nan(x1,x2)
```

### __ADD__

```python
x1 = tf.constant([[2,3,4,5,6],[1,2,4,5,0]],dtype=tf.float32)
x2 = tf.constant([7],dtype=tf.float32)
x2_stretched = tf.constant([7,7,7,7,7],dtype=tf.float32)

print(tf.add(x1,x2))
print(tf.add(x1,x2_stretched))
```

### __SUBTRACT__

```python
x1 = tf.constant([[2,3,4,5,6],[1,2,4,5,0]],dtype=tf.float32)
x2 = tf.constant([7],dtype=tf.float32)
x2_stretched = tf.constant([7,7,7,7,7],dtype=tf.float32)

print(tf.subtract(x1,x2))
print(tf.subtract(x1,x2_stretched))
```

### __MULTIPLY__

```python
x1 = tf.constant([[2,3,4,5,8,9]],dtype=tf.int32)
x2 = tf.constant([[7],[5],[3]],dtype=tf.int32)


print(tf.multiply(x1,x2))
```

### __ARGMAX__

```python
x1 = tf.constant([
    [2,0,-3],
    [3,5,1],                     
    [-9,6,7]
    ])
print(tf.argmax(x1,axis=0))

x2 = tf.constant([2,3,4,55,8,9])
print(tf.argmax(x2))
```

### __ARGMIN__

```python
x1 = tf.constant([
    [2,10,-3],
    [3,5,1],                     
    [-9,6,7]
    ])
print(tf.argmin(x1,axis=1))

x2 = tf.constant([2,3,4,55,1,9])
print(tf.argmin(x2))
```

### __POW__

```python
x1 = tf.constant([[2,3],[1,8]])
x2 = tf.constant([x1[2,1],[4,2]])
tf.pow(x1,x2)
```

### __REDUCE__

```python
t = tf.constant([
    [2,10,-3],
    [3,5,1],                     
    [-9,6,7]
    ])

print(tf.reduce_sum(t,axis=0,keepdims=[1,3],name=None))
print(tf.reduce_max(t))
print(tf.reduce_min(t))
print(tf.reduce_prod(t))
print(tf.reduce_mean(t))
```

### __EXPAND DIMS__

```python
t = tf.constant([
    [2,10,-3],
    [3,5,1],                     
    [-9,6,7]
    ])

print(t.shape)
print(tf.expand_dims(t,axis=0).shape)
```

### __SQUEEZE__

```python
t = tf.constant([

    [
        [2,5,1],[6,7,9]
    ],
    [
        [-9,2,-7],[8,2,6]            
    ],
    [
        [6,5,2],[-1,0,3]
    ]
])
print(t.shape)
expand = tf.expand_dims(t,axis=0)
print(expand.shape)
squeeze = tf.squeeze(expand,axis=0)
print(squeeze)
```

### __RESHAPE__

```python
t = tf.constant([
    [2,10,-3],
    [3,5,1],                     
    [-9,6,7]
    ])

print(tf.reshape(t,[9]))
print(tf.reshape(t,[3,1,3]))
```

### __CONCAT__

```python
x1 = np.array([[2,3],
                  [1,8]])
x2 = np.array([[2,1],
                  [4,2]])

print(tf.concat([x1,x2], axis=0))
print(tf.concat([x1,x2], axis=1))
```

### __NP EINSUM__

```python
x1 = np.array([[2,3],
                  [1,8]])
x2 = np.array([[2,1],
                  [4,2]])

print("MATRICS MULTIPLICATION USING MATMUL :\n")
print(np.matmul(x1,x2),"\n")

print("MATRICS MULTIPLICATION USING EINSUM :\n")
print(np.einsum('ij,jk -> ik',x1,x2),"\n")

print("ELEMENT WISE MULTIPLICATION USING HARDAMOND :\n")
print(x1*x2,"\n")

print("ELEMENT WISE MULTIPLICATION USING EINSUM :\n")
print(np.einsum('ij,ij -> ij',x1,x2),"\n")
```

### __NP TRANSPOSE__

```python
A = np.array([[2,3],
                  [1,8]])

print("TRANSPOSE USING NUMPY :\n")
print(A.T,"\n")

print("TRANSPOSE USING EINSUM :\n")
print(np.einsum('ij->ji',A))
```

### __SPARSE__

### __STRINGS__
