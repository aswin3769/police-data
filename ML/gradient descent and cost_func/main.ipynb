```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def gradient_descent(x,y):
    m_curr = b_curr =0
    iter = 500
    n = len(x)
    learning_rate = 0.02
    
    m_vals = []
    b_vals = []
    cost_vals = []

    for i in range(iter):
        y_predicted = m_curr * x + b_curr
        cost = (1/n) * sum([val**2 for val in (y - y_predicted)])
        md = -(2/n) * sum(x*(y - y_predicted))
        bd = -(2/n) * sum((y - y_predicted))

        m_curr = m_curr - learning_rate * md
        b_curr = b_curr - learning_rate * bd

        print(f'iteration {i+1} : m - {m_curr} , b - {b_curr} , cost - {cost}')

        m_vals.append(m_curr)
        b_vals.append(b_curr)
        cost_vals.append(cost)

        plt.scatter(i , cost,color='red' )

   
    plt.figure(figsize=(18, 5))

    # m vs Iteration
    plt.subplot(1, 4, 1)
    plt.plot(range(1, iter+1), m_vals, color='blue')
    plt.xlabel("Iteration")
    plt.ylabel("m (slope)")
    plt.title("m vs Iteration")

    # b vs Iteration
    plt.subplot(1, 4, 2)
    plt.plot(range(1, iter+1), b_vals, color='green')
    plt.xlabel("Iteration")
    plt.ylabel("b (intercept)")
    plt.title("b vs Iteration")

    # Data + Final Regression Line
    plt.subplot(1, 4, 3)
    plt.scatter(x, y, color='red', label='Data Points')
    y_final = m_curr * x + b_curr
    plt.plot(x, y_final, color='black', label='Final Regression Line')
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title("Final Regression Line")
    plt.legend()

    # Gradient Descent 
    plt.subplot(1, 4, 4)

    # cost function: keep m fixed at final value, vary b
    b_vals_range = np.linspace(b_curr - 1, b_curr + 1, 100)
    costs = []
    for b in b_vals_range:
        y_pred = m_curr * x + b
        cost = (1/n) * sum((y - y_pred)**2)
        costs.append(cost)
    
    plt.plot(b_vals_range, costs, color='green', label="Sum of Squared Residuals")

    # Current point
    current_cost = (1/n) * sum((y - (m_curr * x + b_curr))**2)
    plt.scatter(b_curr, current_cost, color='red', marker='|', s=500, label="Current Point")

    # Tangent line at b_curr
    slope = -(2/n) * sum(y - (m_curr * x + b_curr))
    tangent_line = slope * (b_vals_range - b_curr) + current_cost  # m * (x - x0) + y0
    plt.plot(b_vals_range, tangent_line, color='red', linestyle='--', label='Tangent')

    # Final minimum point
    plt.scatter(b_curr, current_cost, color='green', marker='X', s=120, label='Minimum')

    plt.xlabel("Intercept (b)")
    plt.ylabel("Cost")
    plt.title("Cost vs Intercept")
    plt.legend()
    plt.grid(True)

    plt.tight_layout()
    plt.show()

        

```

```python
x = np.array([1,2,3,4,5])
y = np.array([5,7,9,11,13])

gradient_descent(x,y)
```
