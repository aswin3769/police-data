```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
data = {'area':[2600,3000,3200,3600,4000],'bedrooms':[3.,4.,np.nan,3.,5.],'age':[20,15,18,30,8],'price':[550000,565000,610000,595000,760000]}
df = pd.DataFrame(data)
df
```

```python
df.bedrooms.fillna(df.bedrooms.mean()//1,inplace=True)
```

```python
df
```

```python
model = LinearRegression()
model.fit(df[['area','bedrooms','age']],df.price)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
predicted_prices = model.predict(df[['area','bedrooms','age']])
```

```python
plt.scatter(df['price'], predicted_prices)
plt.plot([min(df.price), max(df.price)], [min(df.price), max(df.price)], color='red')  # perfect prediction line
plt.xlabel('Actual Price')
plt.ylabel('Predicted Price')
plt.title('Actual vs Predicted Prices')
plt.grid(True)
plt.show()
```
